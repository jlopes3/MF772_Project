```python
import pandas as pd
import numpy as np
import yfinance as yf
```

```python
ief = yf.download("IEF", start="2017-01-01", end="2024-12-31", interval="1mo")

# If columns are MultiIndex like ('Price','IEF'), ('Close','IEF'), ...
ief.columns = ief.columns.get_level_values(0)  # keep 'Price','Close','High',...

# Make sure index is datetime
ief.index = pd.to_datetime(ief.index)

# Sort just in case
ief = ief.sort_index()

# Use 'Price' (could also use 'Close', they look very similar monthly)
ief["ret_T"] = ief["Close"].pct_change()

# Create year-month period to match your bond portfolios
ief["ym"] = ief.index.to_period("M")

# Keep only what we need, drop first NaN return
ief_m = (
    ief[["ym", "ret_T"]]
    .dropna(subset=["ret_T"])
    .set_index("ym")
)

ief_m.to_csv('treasury_data.csv')
```
