```python
import pandas as pd
import numpy as np
```

```python

df = pd.read_csv("SecurityData.csv")
df.columns = df.columns.str.strip().str.lower()
```

```python
date_cols = ['date', 'maturity_date', 'next_call_date']
for c in date_cols:
    if c in df.columns:
        df[c] = pd.to_datetime(df[c], errors='coerce')
```

```python

num_cols = ['spread', 'duration', 'closing_price', 'ytm', 'current_yield', 'coupon_rate']
for c in num_cols:
    if c in df.columns:
        df[c] = pd.to_numeric(df[c], errors='coerce')
```

```python
df = df[df['asset_type'].str.lower().isin(['credit', 'corporate', 'corp'])]
```

```python
df['spread'] = df['spread'] / 10000
```

```python
df = df.dropna(subset=['spread', 'duration', 'closing_price'])
```

```python
df = df[(df['spread'] > 0) & (df['duration'] > 0)]
```

```python
if 'date' in df.columns and 'maturity_date' in df.columns:
    df['ttm'] = (df['maturity_date'] - df['date']).dt.days / 365.25
    df = df[df['ttm'] > 0]
```

```python
df['dts'] = df['spread'] * df['duration']
```
